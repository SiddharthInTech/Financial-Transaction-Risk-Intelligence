```python
balance_columns = [
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest"
]
```

```python
zero_by_type = (
    df.groupby("type")[balance_columns]
      .apply(lambda x: (x == 0).mean() * 100)
      .round(2)
)

print("\nPercentage of Zero Balances by Transaction Type:")
print(zero_by_type)
```
