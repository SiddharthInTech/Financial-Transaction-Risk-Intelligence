```python
import pandas as pd
```

```python
# -----------------------------------------
# STEP 1: Loading PaySim dataset
# -----------------------------------------

df = pd.read_csv("../../data/raw//PS_20174392719_1491204439457_log.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# -----------------------------------------
# PHASE 1.4: DATA CLEANING
# STEP 1: Validate Raw Dataset
# -----------------------------------------

print("\n--- Data Validation ---")

# Check missing values
missing_values = df.isnull().sum().sum()
print("Total missing values:", missing_values)

# Check duplicate rows
duplicate_rows = df.duplicated().sum()
print("Total duplicate rows:", duplicate_rows)

# Check negative financial values
financial_columns = [
    "amount",
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest"
]

negative_values = df[financial_columns].lt(0).sum().sum()
print("Total negative financial values:", negative_values)

# Check target values
print("\nTarget distribution:")
print(df["isFraud"].value_counts())

# Final validation
if missing_values == 0 and duplicate_rows == 0 and negative_values == 0:
    print("\nRaw dataset validation: PASSED")
else:
    print("\nRaw dataset validation: REVIEW REQUIRED")
```

```python
# -----------------------------------------
# STEP 2: Remove Existing Fraud Flag
# -----------------------------------------

# Keep a copy of the original structure before modeling
clean_df = df.copy()

# Remove PaySim's existing fraud flag from the primary dataset
clean_df = clean_df.drop(columns=["isFlaggedFraud"])

print("\nAfter Removing isFlaggedFraud:")
print("Shape:", clean_df.shape)

print("\nRemaining Columns:")
print(clean_df.columns.tolist())
```

```python
# -----------------------------------------
# STEP 3: Save Cleaned Dataset
# -----------------------------------------

output_path = "../../data/processed/paysim_clean.csv"

clean_df.to_csv(
    output_path,
    index=False
)

print("\nCleaned dataset saved successfully!")
print("File:", output_path)
print("Shape:", clean_df.shape)
```
